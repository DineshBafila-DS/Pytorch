# Scalar 
A tensor with dimension 0 is called scalar

```python
scalar=torch.tensor(2)
print(scalar)
print(f"dimesnion of this tensor is {scalar.ndim}")
```

# Vector
A Tensor with dimesnion 1 is called vector

```python
vector= torch.tensor([0,1])
print(vector)
print(f"dimesnion of this tensor is {vector.ndim}")
```

```python
matrix= torch.tensor([[0,1],
                      [1,2]])
print(matrix)
print(f"dimesnion of this tensor is {matrix.ndim}")
```

# Tensor
Anything bigger than 2 dimesnion is called Tensor

```python
tensor=torch.tensor([[[0,1],
                      [1,2],
                      [2,3]]])
print(tensor)
print(f"Dimensio of this tensor is {tensor.ndim}")
```

```python
print(tensor.shape)
```

# Indexing

```python
print(tensor)
print(tensor[0])
print(tensor[0][0])
print(tensor[0][1])
print(tensor[0][2])
print(tensor[0][2][1])
print(tensor[:,:,0])
```

# Random Tensor

```python
random_tensor=torch.rand(size=(3,4))
print(random_tensor)
print(random_tensor.ndim)
print(random_tensor.shape)
```

# Application of Random Tensor
Random tensor are widely used in random weight initialization

```python
#Example
input=30
hidden=100
weight_tensor=torch.rand(size=(input,hidden))
print(weight_tensor)
print(weight_tensor.ndim)
print(weight_tensor.shape)
```

# Tensor Manipulation

```python
tensor_t=torch.tensor([1,2,3,4,5])
```

```python
##  Addition
print(tensor_t+10) #immutable
print(tensor_t)
#or
print(torch.add(tensor_t,10)) #immutable
print(tensor_t)
```

```python
##  Substraction
print(tensor_t-10) #immutable
print(tensor_t)
#or
print(torch.sub(tensor_t,10)) #immutable
print(tensor_t)
```

```python
##  multiplication
print(tensor_t*10) #immutable
print(tensor_t)
#or
print(torch.mul(tensor_t,10)) #immutable
print(tensor_t)
```

```python
##  Division
print(tensor_t/10) #immutable
print(tensor_t)
#or
print(torch.div(tensor_t,10)) #immutable
print(tensor_t)
```

# Dot Product or Matrix Multiplication

```python
tensor_a=torch.tensor([[1,2],
                       [2,3],
                       [3,4]])
tensor_b=torch.tensor([[1,2],
                       [2,3],
                       [3,4]])
print(tensor_a.shape)
print(tensor_b.shape)
#take the transpose of first tensor
print(torch.matmul(tensor_a,tensor_b.T))
```

# Aggregation

```python
torch_agg=torch.tensor([1,2,3,4,100,300,400,-100])
print(f"minimum value in this tensor{torch_agg.min()}")
print(f"maxiimum value in this tensor{torch_agg.max()}")
print(f"mean value in this tensor {torch_agg.float().mean()}")
```
